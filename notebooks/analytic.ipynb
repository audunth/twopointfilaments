# Analytic computations for the two-point model

```python
import sympy as s
```

## Isothermal, diffusion+constant convection

```python
r,P,c = s.symbols('r P c',real=True, positive=True)
n_t = s.Function('n_t')
Δ = (s.sqrt(P)/2+s.sqrt(P/4+1))**(-2)
```

```python
eqd = Δ*s.Derivative(n_t(r),(r,2))-s.sqrt(P*Δ)*s.Derivative(n_t(r),(r,1))-n_t(r)/c-s.exp(-r)
s.dsolve(eqd,n_t(r),ics={n_t(0):s.Rational(1/2), n_t(s.oo): 0},hint='best')
```
